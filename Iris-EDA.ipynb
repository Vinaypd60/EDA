```python
import seaborn as sns
import matplotlib.pyplot as plt

# Base plot
ax = sns.boxplot(x="Species", y="PetalLengthCm", data=iris)

# Overlay stripplot
sns.stripplot(x="Species", y="PetalLengthCm", data=iris, 
              jitter=True, edgecolor="gray", ax=ax)

# Recolor each box
colors = ["green", "red", "yellow"]   # one per species
for i, artist in enumerate(ax.artists):
    artist.set_facecolor(colors[i])
    artist.set_edgecolor("black")

plt.title("Petal Length by Species", fontsize=14)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select only numeric columns
numeric_df = iris.select_dtypes(include=['float64','int64'])

plt.figure(figsize=(10,7))
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',   # you can change to 'cubehelix', 'viridis', etc.
    linewidths=1,
    linecolor='black',
    square=True,
    vmin=-1,
    vmax=1,
    cbar_kws={"orientation": "vertical"},
    cbar=True
)

plt.title("Correlation Heatmap (Numeric Features)", fontsize=16)
plt.show()
```

```python
sub = iris[iris['Species'] == 'Iris-setosa']

# 2D KDE plot of Sepal Length vs Sepal Width
sns.kdeplot(
    data=sub,
    x="SepalLengthCm",
    y="SepalWidthCm",
    cmap="plasma",
    fill=True,        # replaces shade=True
    thresh=0,         # ensures full shading
    levels=50         # number of contour levels
)

plt.title("Iris-setosa Sepal Length vs Width", fontsize=14)
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.show()
```
